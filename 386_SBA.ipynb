```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## Section One

```python
# 1.Read data from CSV file
raw_df = pd.read_csv('data/woocommerce-product-export.csv')
df = raw_df.copy()
```

```python
# 2. Show a concise summary of the columns using the info() method.
print(df.info())
```

```python
# 3. Show a summary of statistics pertaining to the columns. 
print(df.describe())
```

```python
# 4. Print the first five rows by default.
print(df.head())
```

```python
# 5. Print the last five rows by default.
print(df.tail())
```

```python
# 6. Print  the "total_profit" and “month_number” columns only.
print(df[["total_profit", "month_number"]])
```

```python
# 7.Read the total profit of all months and show it using the Bar plot.
plt.figure(figsize=(6, 5)) #played with the figure size a lot to make it look like SBA image but couldn't get it exactly right

plt.bar(df["month_number"], 
df["total_profit"])

plt.xlabel("Month number")
plt.ylabel("Total profit")
plt.title("Company profit per Month")

plt.xlim(0, 13)
plt.ylim(0, 500000)

plt.grid(True, linestyle='--', linewidth=0.8, alpha=0.7)
plt.show()
```

```python
# 8. Read the total profit of all of the months, and show the line plot.
plt.figure(figsize=(6, 5))

plt.plot(df["month_number"], df["total_profit"], 
marker='o',
markerfacecolor='black', 
linewidth=3, 
linestyle='--', 
color='red',
label='Profit data of last year' 
)

plt.xlabel("Month number")
plt.ylabel("Total profit")
plt.title("Company profit per month")

plt.xticks(df["month_number"])

plt.legend(loc='lower right')
plt.show()
```

```python
# 9. Print all of the product sales data and show it using a multi-line plot.
plt.figure(figsize=(6, 5))

products = df.columns[1:7]
#Created a list tomake legend names match 
labels = [
    "Face cream Sales Data",
    "Face Wash Sales Data",
    "Toothpaste Sales Data",
    "Bathingsoap Sales Data",
    "Shampoo Sales Data",
    "Moisturizer Sales Data"
]

for product, label in zip(products, labels):
    plt.plot(df["month_number"], df[product],
             marker='o',
             label=label)

plt.xlabel("Month Number")
plt.ylabel("Sales units in number")
plt.title("Sales Data")

plt.ylim(0, 18000) # Added to make y-axis match

plt.xticks(df["month_number"])
plt.legend(loc="upper left")

plt.show()


```

```python
# 10. Read “bathingsoap” sales data for each month and show it using a scatter plot.
plt.figure(figsize=(6, 5))

plt.scatter(
    df["month_number"], 
    df["bathingsoap"], 
    marker='o', 
    color='blue',
    label='bathing soap sales data')

plt.xlabel("Month Number")
plt.ylabel("Number of units Sold")
plt.title("bathingsoap Sales data")


plt.xticks(df["month_number"])

plt.grid(True, linestyle='--', linewidth=0.5)
plt.legend(loc="upper left")
plt.show()
```

```python
# 1. A line chart of date versus temperature

date=["25/12","26/12","27/12"]
temp=[8.5,10.5,6.8]

plt.figure(figsize=(6, 5))

plt.plot(
    date, 
    temp, 
    marker='*', 
    markersize=10, 
    color='green', 
    label='Temperature'
    )
plt.xlabel("Date")
plt.ylabel("Temperature")
plt.title("Date-wise Temperature")

plt.grid(True, linestyle='--', linewidth=0.5)
plt.legend(loc="upper left")

plt.show()

```

```python
# 2. Show the average weight against the average height

height=[121.9,124.5,129.5,134.6,139.7,147.3,152.4,157.5,162.6]
weight=[19.7,21.3,23.5,25.9,28.5,32.1,35.7,39.6,43.2]

plt.figure(figsize=(6, 5))

plt.plot(
    weight, 
    height, 
    marker='o', 
    markersize=10, 
    color='green', 
    label='Weight vs Height',
    linestyle='-.'
)
plt.xlabel("Weight in kg")
plt.ylabel("Height in cm")
plt.title("Average weight with respect to average height")

plt.legend(loc="lower right")

plt.show()


```
